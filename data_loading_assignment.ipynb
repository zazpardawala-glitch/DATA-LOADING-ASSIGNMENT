Q9) Q9. Loading Strategy Selection:

Assume this dataset represents daily sales data.

a) Should a Full Load or Incremental Load be used?

b) Justify your choice.

ans)
a) Load Type Choice:

Incremental Load should be used.

b) Justification:

1. NATURE OF THE DATA:

The dataset represents daily sales transactions.

New records are added each day, while historical records typically remain unchanged (except for rare corrections).

2. EFFECIENCY:

Full Load means reloading all historical sales data every day. That’s inefficient and costly as the dataset grows.

Incremental Load only processes new or changed records since the last load, reducing processing time and resource usage.

3. BUSINESS CONTINUITY:

Incremental loading ensures dashboards and reports stay up-to-date quickly without long downtime.

Faster loads mean near real-time reporting is possible.

4. DATA QUALITY HANDLING:

With incremental loads, you can still apply validations (duplicates, nulls, format issues) on the new batch before merging it into the warehouse.

Historical data remains stable, so you don’t risk reintroducing errors by reloading everything.

5. EXCEPTION HANDLING:

If corrections are needed (e.g., fixing a wrong Sales_Amount), you can design the incremental process to detect updates (using Order_Date + Order_ID as keys, or CDC/change tracking).

This balances efficiency with accuracy.

When Full Load might be justified:

If the dataset is small and easy to reload daily.

If there are frequent retroactive changes to historical sales (e.g., backdated orders, corrections).

If the system lacks proper change tracking or incremental logic.

But in a real-world daily sales pipeline, Incremental Load is the standard choice because it’s scalable, efficient, and business-friendly.
